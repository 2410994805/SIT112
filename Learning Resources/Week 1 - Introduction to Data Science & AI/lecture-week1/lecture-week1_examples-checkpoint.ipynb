## Get the data

```python
# Import the pandas library with an alias "pd"
import pandas as pd

# Assign the URL of a CSV file containing 2016 US presidential election polls to a variable
poll_url = 'https://vincentarelbundock.github.io/Rdatasets/csv/dslabs/polls_us_election_2016.csv'

# Use pandas to read the CSV file from the URL and assign it to a variable called "polls"
polls = pd.read_csv(poll_url)

# Display the "polls" dataframe 
display(polls)
```

## Sort the data

```python
# Sort the "polls" dataframe by the "startdate" column in descending order, with the changes applied in place
polls.sort_values('startdate', ascending=False, inplace=True)

# Print the first 5 rows of the sorted "polls" dataframe 
display(polls.head())
```

## Use lists, slices, tuples, and dictionary objects

### A list

```python
# Drop the specified columns from the "polls" dataframe and apply the changes in place
polls.drop(columns=['rawpoll_johnson','rawpoll_mcmullin','adjpoll_johnson','adjpoll_mcmullin'], inplace=True)
# Print the first 2 rows of the modified "polls" dataframe to the console
polls.head()
```

### A tuple

```python
# Convert the startdate and enddate columns to datetime objects
polls['startdate'] = pd.to_datetime(polls['startdate'], errors='coerce')
# Create a line plot of the "polls" dataframe with the x-axis limited to the specified range
ax = polls.plot.line(x='startdate',y=['rawpoll_clinton', 'rawpoll_trump'])
# Use a TUPLE for xlim
ax.set_xlim((pd.Timestamp('2016-10-01'),pd.Timestamp('2016-11-06')))
```

### A dictionary

```python
# Rename the columns of the "polls" dataframe using a dictionary of column names
polls.rename(columns={'adjpoll_clinton':'Clinton', 'adjpoll_trump':'Trump'}, inplace=True)
```

### Two slices

```python
# Select a subset of rows and columns from the "polls" dataframe using label indexing
# The rows are selected using a slice object from index 0 to 100 with a step of 10
# The columns are selected using label indexing with a slice object from 'state' to 'grade'
subset = polls.loc[0:100:10,'state':'grade']
subset
```

## How to code a list comprehension

```python
# Generate a list of values for the x-axis ticks using a list comprehension
# The list comprehension generates values from 1900 to 1920 with a step of 2
xticks = [x for x in range(1900,1920,2)]
xticks
```

## How to continue statements

### With implicit continuation

```python
# Sort the "polls" dataframe by two columns in descending order using the sort_values method
# The first column to sort by is 'state', and the second is 'startdate'
# The resulting sorted dataframe is stored in place (in the same variable) using inplace=True
polls.sort_values(['state','startdate'], 
                  ascending=False,
                  inplace=True)

# Display the first few rows of the sorted dataframe using the head() method
polls.head()
```

### With explicit continuation

```python
polls.sort_values(['state','startdate'], \
                  ascending=False, \
                  inplace=True)
polls.head()
```

## How to use the Magic Commands

```python
# Use the magic command %time to measure the time it takes to execute the following line of code
%time polls = pd.read_csv(poll_url)

# Display the "polls" dataframe
polls
# CPU times: The total CPU time used by the process. It is divided into two parts: user and system time.
# user: The amount of CPU time used by the code that was executed in user mode (i.e., code that was executed in user mode).
# sys: The amount of CPU time used by the system on behalf of the process (i.e., code that was executed in kernel mode).
# total: The sum of user and sys time.
# Wall time: The elapsed real time (i.e., the time that it took to execute the code in wall-clock time, as opposed to CPU time).
```

```python
%%time
polls = polls.sort_values('startdate', ascending=False)
polls
# The '%%time' magic command prints the amount of time it takes for its following code to run
```
