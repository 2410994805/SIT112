What is the difference between using request.urlretrieve and simply giving a URL to the read_csv method?

This is an excellent question, and it's worth explaining to students because both approaches are valid but serve different purposes.

**Option 1: Read directly from a URL**
import pandas as pd

``` python
url = "https://example.com/data.csv"
df = pd.read_csv(url)
```


✅ One line of code

✅ Fast and convenient

❌ The file is not saved on your computer.

❌ You cannot inspect or reuse the downloaded file later.

**Option 2: Download first with urlretrieve()**

``` python
from urllib import request
import pandas as pd
url = "https://example.com/data.csv"
request.urlretrieve(url, "data.csv")
df = pd.read_csv("data.csv")
```
✅ The file is permanently stored on your computer.

✅ You can inspect it before reading.

✅ You can reuse it later without downloading again.

✅ Useful for ZIP files, Excel files, images, SQLite databases, PDFs, etc.

❌ Requires two steps.

urlretrieve() is a general download tool. It can download any file type, not just CSV files.

## How to work with a zip file on disk

### Download the zip file to disk

````python
from urllib import request

url = "https://github.com/mwaskom/seaborn-data/archive/refs/heads/master.zip"

# Download the ZIP file
request.urlretrieve(url, "seaborn-data.zip") # saves in the current directory
````

### Extract the files and list their names

````python
# Import the ZipFile class from the zipfile module
from zipfile import ZipFile

# Create an empty list to store the names of extracted files
file_names = list() 

# Use the ZipFile class to open the zip file and extract its contents
with ZipFile('seaborn-data.zip', mode='r') as zip:
    zip.extractall('./data')
    # Iterate over each file in the zip file
    for file in zip.infolist():
        # Add the file's name to the list of file names
        file_names.append(file.filename)
        # Print the file's name, compressed size, and actual size
        print(file.filename, file.compress_size, file.file_size)
````

## How to read excel file in a zip file

````python
from zipfile import ZipFile
import pandas as pd

# Path to the ZIP file
file_path = "./data/sample_data.zip"

# List to store file names
file_names = [] # list()

# Extract all files into the 'data' folder
with ZipFile(file_path, mode="r") as zip_ref:
    zip_ref.extractall()
    # Display information about the files in the ZIP archive
    print("\nFiles in the ZIP archive:")
    for file in zip_ref.infolist():
        file_names.append(file.filename)
        print(f"{file.filename} | Compressed: {file.compress_size} bytes | Original: {file.file_size} bytes")


# Read the Excel file from the extracted folder
df = pd.read_excel("./data/" + file_names[0], sheet_name=0) # , sheet_name=0 is the extracted file
print("\n")
df.head()
````

## How to run queries against a database

````python
# Import the sqlite3 module
import sqlite3

# Create a connection object to the DemoJobs.sqlite database
DemoJobs_con = sqlite3.connect('./data/DemoJobs.sqlite')

# Create a cursor object to execute SQL commands on the database
# cursor:  A database object that allows developers to traverse or retrieve rows one by one from a result set. 
# A cursor is a database object that acts as a pointer to the results of a SQL query.
DemoJobs_cur = DemoJobs_con.cursor()

# Use the cursor to execute an SQL command to retrieve the names of all tables in the database
DemoJobs_cur.execute('SELECT name, type FROM sqlite_master WHERE type="table"')

# Fetch all rows returned by the SQL command and print them
DemoJobs_cur.fetchall()
````

## How to use a SQL query to import data into a DataFrame

### Get information about a table

````python
# Execute a SQL PRAGMA command to retrieve information about the columns in the 'DemoJobs' table
# of the SQLite database connected to by 'DemoJobs_cur'
DemoJobs_cur.execute('PRAGMA table_info(Jobs)') # Applications, Jobs, Companies, Applicants

# Fetch all of the rows from the result set as a list of tuples
# Each tuple represents one column in the table, with values for the column ID, name, data type, etc.
result_set = DemoJobs_cur.fetchall()

# Print the result set
print(result_set)
````

### Import the data from a query into a DataFrame

````python
import pandas as pd
# Execute an SQL query to retrieve data from the 'Jobs' table of 
# the SQLite database connected to by 'DemoJobs_con'
# The SELECT statement retrieves specific columns from the table, 
query = '''
        SELECT JobID, JobTitle, Salary
        FROM Jobs
        '''

# Use the pd.read_sql_query() method to execute the query and store 
# the results in a Pandas DataFrame called 'DemoJobs'
# The method takes two arguments: the SQL query to execute, 
# and the database connection to use
DemoJobs = pd.read_sql_query(query, DemoJobs_con)
DemoJobs.head()
````

## How to build DataFrames for the metadata and the data in a Stata file

### Get metadata from a Stata file

````python
import pandas as pd
import pyreadstat
# Set the filename of the Stata file to be read in

stata_filename = 'data/auto.dta'

# Use the pyreadstat library to read in the metadata from the Stata file
df, meta = pyreadstat.read_dta(stata_filename, metadataonly=True) 
# metadataonly=True return empty df

# Print the type of the metadata object
print(type(meta))
df.head()
````

### What the attributes of the metadata container can tell you

````python
print("Number of columns:", meta.number_columns)
print("Number of rows:", meta.number_rows)
````

````python
print("column_labels:", meta.column_labels)
print("column_names:", meta.column_names)
````

### How to build a DataFrame for the column descriptions in the metadata

````python
import pandas as pd

meta_cols = pd.DataFrame(
    data = meta.column_labels, 
    # The data is the column_labels attribute from the meta object
    index = meta.column_names, 
    # The row index is the column_names attribute from the meta object
    columns=['description']  
    # The column name is 'description'
)

meta_cols.head()  # Display the first five rows of the resulting DataFrame
````

### How to import columns of the data into a DataFrame

````python
# Define the file path for the dataset
file_path = 'data/auto.dta'

# Specify the columns to be selected for the data frame
selected_columns = ['make',	'price', 'headroom', 'trunk']

# Read the Stata dataset with read_stata function and select only the specified columns
data = pd.read_stata(file_path, columns=selected_columns)

# Display the last 5 rows of the data frame
data.tail()
````

## How to download a JSON file

````python
# import json module to work with JSON data
import json  
# import request module to retrieve data from URLs
from urllib import request  

# URL for the JSON data file
shots_url = 'https://www.murach.com/python_analysis/shots.json'  
# retrieve the data and save it to a file named 'shots.json'
request.urlretrieve(shots_url, filename='shots.json')  
````

## How to build a DataFrame for the data

````python
# Open the 'shots.json' file using a context manager
# The file is opened in read-only mode and assigned to the variable 'json_data'
with open('shots.json') as json_data:
    # Load the contents of the JSON file into a dictionary called 'shots'
    shots = json.load(json_data)

# Extract the 'rowSet' from the 'resultSets' key in the 'shots' dictionary and assign it to 'allRows'
allRows = shots['resultSets'][0]['rowSet']

# Extract the column headers from the 'headers' key in the 'resultSets' key in the 'shots' dictionary
# Convert the column headers to lowercase using a list comprehension and assign them to 'columnHeaders'
columnHeaders = [x.lower() for x in shots['resultSets'][0]['headers']]

# Create a pandas DataFrame called 'shots' with 'allRows' as the data and 'columnHeaders' as the column names
shots = pd.DataFrame(data=allRows, columns=columnHeaders)

# Display the first few rows of the 'shots' DataFrame using the 'head()' method
shots.head()
````
