Ex:
The user-defined function `'convert_sex'` takes a **row** from the 'workData' DataFrame as **input**.
 The function checks the value of the `'sex'` column in the row, 
 and **returns `'male'` for 1**, **`'female'` for 2**, and `'non-binary'` for **any other value**.
 Apply the `'convert_sex'` function to each row in the `'workData'` DataFrame,
 along the **row axis** `(axis=1)`, which **updates the 'sex' column** with the corresponding converted values.

````python
def convert_sex(row):
    if row.sex == 1: # row.sex means the value in the sex column for the current row.
        return 'male'
    elif row.sex == 2:
        return 'female'
    else:
        return 'non-binary'
    
workData['sex'] = workData.apply(convert_sex, axis=1)
# This line of code displays the updated 'workData' DataFrame using the 'head()' method.
workData.head()
````

Ex:

This function **takes a row as input** and calculates the corresponding season based on the `'game_date'` month.

 If the month > 6, it assigns the season as the concatenation of the `'game_date.year'` and `'game_date.year + 1'`. Otherwise, it assigns the season as the concatenation of the `'game_date.year - 1'` and `'game_date.year'`.
The calculated season is returned as the output.

````python
def get_season(row):
    if row.game_date.month > 6:
        season = f'{row.game_date.year}-{row.game_date.year + 1}'
    else:
        season = f'{row.game_date.year - 1}-{row.game_date.year}'
    return season

# Apply the 'get_season' function to each row in the 'gameData' DataFrame along the specified axis (axis=1),
# and assign the calculated season to a new column called 'season' in 'gameData'.

gameData['season'] = gameData.apply(get_season, axis=1)

# Temporarily set the display options for the 'gameData' DataFrame.
# Set the maximum number of rows to display as 6
with pd.option_context('display.max_rows', 6, 'display.max_columns', None):
    display(gameData)
````

- `'display.max_columns'`, None displays all columns without truncating them.

`stack()` and `unstack()` reshape data between wide and long formats, usually when working with a MultiIndex.

**Wide format**

A wide table stores categories across multiple columns:

| state | Jan | Feb | Mar |
| ----- | --: | --: | --: |
| CA    |  10 |  15 |  12 |
| NY    |   5 |   8 |   7 |


`stack()`: **wide → long**              
```python
long_df = wide_df.stack()
```

| state | month | value |
| ----- | ----- | ----: |
| CA    | Jan   |    10 |
| CA    | Feb   |    15 |
| CA    | Mar   |    12 |
| NY    | Jan   |     5 |
| NY    | Feb   |     8 |
| NY    | Mar   |     7 |


`stack()` moves column labels into the row index. It “stacks” the columns vertically.

To convert the result into a regular DataFrame:
```python
long_df = (
    wide_df
    .stack()
    .reset_index(name='number_of_fires')
)
```
`unstack()`: **long → wide**

Suppose the long data has a MultiIndex consisting of state and month:

```python
wide_df = long_df.unstack()
```

| state | Jan | Feb | Mar |
| ----- | --: | --: | --: |
| CA    |  10 |  15 |  12 |
| NY    |   5 |   8 |   7 |


`unstack()` moves **one row-index level into columns**. It “spreads” the values horizontally.



| Method      | Movement                  | Typical result |
| ----------- | ------------------------- | -------------- |
| `stack()`   | Columns → row index       | Wide → long    |
| `unstack()` | Row-index level → columns | Long → wide    |


You can specify which level to move:
```python
df.stack(level='month')
df.unstack(level='month')
```

If you do not specify level, unstack() uses the **last (innermost) index level** by default.

````python
# Unstack the data in the 'top5_states' DataFrame at the state level.
# The 'level' parameter is set to 'state' to specify that the 
# state column should be used as the new column headers.
top_wide = top5_states.unstack(level='state')
display(top_wide.head(3))
````

````python
print(top_wide.columns)
print(top_wide.columns.names)
print(top_wide.columns.nlevels)
# None appears because the first level of your column MultiIndex does not have a name.
````

You may obtain this structure by unstacking a DataFrame containing multiple value columns:


## How to join DataFrames

```python
DF = pd.concat([df1, df2, df3])
```
By default, it combines them vertically (`axis=0`), meaning their rows are placed one below another:
- `axis=0` is the default and combines rows vertically.
- `ignore_index=True` discards the original row indexes and creates a new index: 0, 1, 2, ....
- The DataFrames should generally have the same column names. If a **column is missing** from one DataFrame, pandas still includes it and fills the missing values with `NaN`.
- The original DataFrames are not modified; a new DataFrame is returned.

````python
# Read the 'shot_cleaned.pkl' file into 'allShotData'.
allShotData = pd.read_pickle('shot_cleaned.pkl')
# Drop unnecessary columns from 'allShotData' 
shots = allShotData.drop(columns=['period','minutes_remaining',
                                  'seconds_remaining','loc_x','loc_y','home_team',
                                  'game_date','shot_attempted_flag','shot_made_flag',
                                  'action_type','visiting_team'])
# Select the first 2 rows from 'shots'
shots1 = shots.head(2)
# Query the 'shots' DataFrame for rows where the 'game_id' = "0020900030"
shots2 = shots.query('game_id == "0020900030"').head(1)
# Query the 'shots' DataFrame for rows where the 'game_id' = "0020900069"
shots3 = shots.query('game_id == "0020900069"').head(1)
# Concatenate the 'shots1', 'shots2', and 'shots3' DataFrames along the row axis (vertically), 
# ignoring the original index.
shots = pd.concat([shots1,shots2,shots3]) # , ignore_index=True
# Set the 'game_id' column as the index of the 'shots' DataFrame.
shots.set_index('game_id', inplace = True)
display(shots)
````

### Inner join in pandas

An inner join keeps only the rows whose joining key **exists in both DataFrames**.

```python
result = left_df.join(
    right_df,
    how='inner',
    lsuffix='_left',
    rsuffix='_right'
)

# _left and _right distinguish the two columns that were both originally named score.
```
- `lsuffix`: added to **overlapping column** names from the **left DataFrame**.
- `rsuffix`: added to **overlapping column** names from the **right DataFrame**.

For example:

```python
left_df
```

| index | score |
| ----- | ----: |
| A     |    80 |
| B     |    90 |

```python
right_df
```

| index | score |
| ----- | ----: |
| A     |    75 |
| C     |    85 |


After the **inner join**:

| index | score_left | score_right |
| ----- | ---------: | ----------: |
| A     |         80 |          75 |


`DataFrame.join()` matches **indexes** by default.

````python
shots_joined = shots.join(points_by_game, how='inner')
display(shots,points_by_game,shots_joined)
````

````python
# Create an independent copy of the points_by_game DataFrame 
points_by_game2 = points_by_game.copy(deep=True)
# Add a new column 'player_name' to the copied DataFrame and set all values to 'Steph Curry'
points_by_game2['player_name'] = 'Steph Curry'
# Display the resulting DataFrame
points_by_game2
````

This is because `deep=True` is the `default`.

`copy(deep=True)` creates a **new DataFrame** with a **separate copy** of its data and index. Changing ordinary values in the copy does not change the original.
`copy(deep=False)` creates a **shallow copy** that **shares underlying data with the original**.

**Left join** 

Keeps every left row, **whether it has a match or not** and fills **unmatched** values with `NaN`.

````python
# Join the 'shots' and 'points_by_game2' DataFrames on their common columns,
# with left and right suffixes '_1' and '_2' respectively, using a left join
shots_joined = shots.join(points_by_game2, lsuffix='_1', rsuffix='_2', how='left')
display(shots,points_by_game2,shots_joined)
````

## outer join
An outer join keeps **every key from both** DataFrames and fills **unmatched** values with `NaN`.

````python
# Join the 'shots' and 'points_by_game2' DataFrames on their common columns,
# with left and right suffixes '_1' and '_2' respectively, using an outer join
shots_joined_outer = shots.join(points_by_game2, lsuffix='_1', rsuffix='_2', how='outer')
display(shots,points_by_game2,shots_joined_outer)
````

## Merge

Both `join()` and `merge()` combine DataFrames by matching related rows. The main difference is the default **joining key**.


| Feature              | `join()`             | `merge()`                           |
| -------------------- | -------------------- | ----------------------------------- |
| Default matching key | DataFrame **indexes**    | One or more **columns**                 |
| Called on            | A DataFrame          | Usually `pd.merge()` or a DataFrame |
| Overlapping columns  | `lsuffix`, `rsuffix` | `suffixes=('_x', '_y')`             |
| Best suited for      | Index-based matching | Database-style column matching      |

````python
# Reset the index of the 'shots' DataFrame
shots2 = shots.reset_index() # to undo what we have done on these DFs.
# Reset the index of the 'points_by_game' DataFrame
points_by_game2 = points_by_game.reset_index()
# Merge the 'shots2' and 'points_by_game2' DataFrames on the 'game_id' column using a left merge
shots_merged = shots2.merge(points_by_game2, on='game_id', how='left')
# Display the resulting merged DataFrame
display(shots2, points_by_game2, shots_merged)
````

````python
df = shots.copy(deep=True)  # Create a deep copy of the 'shots' DataFrame and assign it to 'df'
display(df.head(2))

dfSlice = df.query('game_id == "0020900015"')  # Select rows from 'df' where 'game_id' is "0020900015" and assign it to 'dfSlice'
dfSlice.loc[:,'player_name'] = 'Curry'  # Update the 'player_name' column in 'dfSlice' to "Curry"
display(dfSlice.head(2))
````

````python
display(df.head(2))
````

#### How to use the copy() method to stop the warning message

````python
dfFixed = df.query('game_id == "0020900015"').copy()  
dfFixed.loc[:, 'player_name'] = 'Curry'
display(dfSlice.head(2))
````

````python
display(df.head(2))
````

### Generates the warning and corrupts the data

````python
df = shots.copy(deep=True)  # Create a deep copy of the 'shots' DataFrame and store it in 'df'
dfSlice = df.loc['0020900015', :]  # Slice the 'df' DataFrame to get rows with index '0020900015' and store it in 'dfSlice'
dfSlice.loc[:, 'player_name'] = 'Curry'  # Update the 'player_name' column of 'dfSlice' with the value 'Curry'
display(dfSlice.head(2))
````

````python
display(df.head(2))
````

#### How to fix this code

````python
df = shots.copy(deep=True)  
dfFixed = df.loc['0020900015', :].copy()  
dfFixed.loc[:, 'player_name'] = 'Curry' 
display(dfSlice.head(2))
````

````python
display(df.head(2))
````

## When the SettingWithCopyWarning isn’t given

````python
df = shots.copy(deep=True)  # Create a deep copy of 'shots' DataFrame and assign it to 'df'
dfSlice = df  # Create a reference of 'df' DataFrame and assign it to 'dfSlice'
dfSlice.loc[:,'player_name'] = 'Curry'  # Modify 'player_name' column of 'dfSlice'
dfSlice.head(2)  # Display the first 2 rows of 'dfSlice'
````

````python
df.head(2)  # Display the first 2 rows of 'df'
````

### How to fix this code

````python
df = shots.copy(deep=True)  # Create a deep copy of the 'shots' DataFrame and assign it to 'df'
dfFixed = df.copy()  # Create a deep copy of the 'df' DataFrame and assign it to 'dfFixed'
dfFixed.loc[:,'player_name'] = 'Curry'  # Set the value 'Curry' to the 'player_name' column of 'dfFixed'
dfFixed.head(2)  # Display the first 2 rows of the 'dfFixed' DataFrame
````
